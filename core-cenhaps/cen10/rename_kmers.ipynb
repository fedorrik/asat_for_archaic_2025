```python
import pandas as pd
import numpy as np
```

```python
def renamer(table_path):
    df = pd.read_csv(table_path, sep='\t').set_index('id')
    renamer = pd.read_csv('inputs/cenhap_renamer.tsv', sep='\t')
    renamer_dict = dict(zip(renamer['old_name'], renamer['new_name']))
    df = df.rename(renamer_dict, axis=1)
    df.to_csv(table_path, sep='\t')
```

```python
renamer('inputs/archaic_counts.tsv')
renamer('inputs/modern_counts.tsv')
```
